```python
from outils.copy_values import check_formula


class shapes(object):
    
    def __init__(self, tuple):
        self.left = tuple[0]
        self.top = tuple[1]
        self.right = tuple[2]
        self.bottom = tuple[3]
        self.rows = self.bottom - self.top + 1
        self.cols = self.right - self.left + 1

    def __str__(self):
        return f"({self.left}, {self.top}, {self.right}, {self.bottom})"
    
    def build_values(self, sheet):

        topleft = sheet.cell(row=self.top, column=self.left)

        if self.isonecell():
            if check_formula(topleft):
                return None
            else:
                return [[topleft.value]]
        else:
            values = []
            found_formula = False

            for row in range(self.rows):
                row_values = []
                for col in range(self.cols):
                    cell = sheet.cell(row=row, column=col)
                    if check_formula(cell):
                        found_formula = True
                        break
                    else:
                        row_values.append(cell.value)
                if found_formula:
                    break
                values.append(row_values)
            return values





shapes1 = shapes(onecell)
shapes2 = shapes(range)

shapes1.rows, shapes1.cols
shapes2.rows, shapes2.cols

shapes1.top
```

```python
def letter_to_number(col):
    col = col.upper()
    n = 0
    for c in col:
        if not ("A" <= c <= "Z"):
            raise ValueError(f"Invalid column letter: {col}")
        n = n * 26 + (ord(c) - ord("A") + 1)
    return n

# examples
letter_to_number("A")   # 1
```

```python
references_old = []
for NR in NR_old:
    references_old.append(NR_old[NR].attr_text.split("!")[1])

shapes_old = []
for i in range(len(list_NR_old)):
    a = references_old[i].split("$")[1:]
    for j in range(len(a)):
        if j % 2 != 1:
            a[j] = letter_to_number(a[j])
    if( len(a) == 4 ):
        a[1] = a[1][:-1]
    a = [int(ciao) for ciao in a]
    shapes_old.append(a)

df_old = pd.DataFrame({
    "name_ranges": list_NR_old,
    "cell_references": references_old,
    "cell_shapes": shapes_old,
})
df_old
```

```python
# Extracting cell values from old workbook based on cell references and shapes

sheet_old = with_data["Sheet1"]

cell_values = []

for i in range(len(df_old)):
    if len(df_old["cell_shapes"][i]) == 2:
        cell_values.append([sheet_old[df_old["cell_references"][i]].value])
    else:
        cell_values_temp = []
        for row in sheet_old.iter_rows( min_row=df_old["cell_shapes"][i][1],
                                        max_row=df_old["cell_shapes"][i][3],
                                        min_col=df_old["cell_shapes"][i][0],
                                        max_col=df_old["cell_shapes"][i][2] ):
            cell_values_temp.append([])
            for col in row:
                cell_values_temp[-1].append(col.value)
        cell_values.append(cell_values_temp)

df_old["cell_values"] = cell_values
df_old
```

```python
def month_name_to_number(month_name):
    # Dictionary mapping month names to numbers
    months = {
        "January": 1, "February": 2, "March": 3, "April": 4,
        "May": 5, "June": 6, "July": 7, "August": 8,
        "September": 9, "October": 10, "November": 11, "December": 12
    }
    # Return the corresponding number or None if not found
    return months.get(month_name.capitalize(), None)

print(month_name_to_number("March"))  # Example usage
```
